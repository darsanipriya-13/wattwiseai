```python

# ================================================================
# WATTWISE AI
# HOUSEHOLD ELECTRICITY CONSUMPTION FORECASTING
# COMPLETE GOOGLE COLAB PROJECT
# IMPORT -> DATASET -> LSTM -> EVALUATION -> WEBSITE -> LAUNCH
# ================================================================

# ========================= 1. IMPORTS ============================

import os
import io
import zipfile
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import gradio as gr

from google.colab import files
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

np.random.seed(42)
tf.random.set_seed(42)

print("⚡ WATTWISE AI INITIALIZING...")
print("TensorFlow:", tf.__version__)


# ========================= 2. CONFIGURATION ======================

FEATURES = [
    "Global_active_power",
    "Global_reactive_power",
    "Voltage",
    "Global_intensity",
    "Sub_metering_1",
    "Sub_metering_2",
    "Sub_metering_3"
]

TARGET = "Global_active_power"
LOOKBACK = 60
RECORDS = 10000
EPOCHS = 12


# ========================= 3. DATASET UPLOAD =====================

print("\nUpload your Kaggle dataset.")
print("Supported: household_power_consumption.txt or ZIP")

uploaded = files.upload()

if not uploaded:
    raise ValueError("No dataset uploaded.")

filename = next(iter(uploaded))
raw_bytes = uploaded[filename]

# Support Kaggle ZIP and direct TXT/CSV uploads

if filename.lower().endswith(".zip"):

    with zipfile.ZipFile(io.BytesIO(raw_bytes)) as archive:

        candidates = [
            n for n in archive.namelist()
            if n.lower().endswith((".txt", ".csv"))
            and not n.startswith("__MACOSX/")
        ]

        if not candidates:
            raise ValueError("No TXT or CSV dataset found in ZIP.")

        # Prefer original household power dataset
        selected_file = next(
            (n for n in candidates
             if "household_power_consumption" in n.lower()),
            candidates[0]
        )

        raw_bytes = archive.read(selected_file)

elif not filename.lower().endswith((".txt", ".csv")):
    raise ValueError("Please upload a TXT, CSV or ZIP dataset.")

df = pd.read_csv(
    io.BytesIO(raw_bytes),
    sep=";",
    na_values=["?"],
    low_memory=False
)

df.columns = df.columns.str.strip()

print("\nOriginal Dataset Shape:", df.shape)
print(df.head())


# ========================= 4. DATA PREPROCESSING =================

required = ["Date", "Time"] + FEATURES

missing_columns = [
    col for col in required if col not in df.columns
]

if missing_columns:
    raise ValueError(
        "Required columns missing: " + str(missing_columns)
    )

df["Datetime"] = pd.to_datetime(
    df["Date"].astype(str) + " " + df["Time"].astype(str),
    dayfirst=True,
    errors="coerce"
)

df = df.dropna(subset=["Datetime"])
df = df.sort_values("Datetime").reset_index(drop=True)

for col in FEATURES:
    df[col] = pd.to_numeric(df[col], errors="coerce")

if len(df) < RECORDS:
    raise ValueError("Dataset has fewer than 10,000 records.")

# Select a contiguous 10,000-record window
# with the fewest missing feature values.

missing_counts = df[FEATURES].isna().sum(axis=1)

rolling_missing = missing_counts.rolling(
    RECORDS,
    min_periods=RECORDS
).sum()

best_end = int(rolling_missing.idxmin())
best_start = best_end - RECORDS + 1

data = df.iloc[
    best_start:best_end + 1
].copy().reset_index(drop=True)

print("\nSelected Records:", len(data))
print("From:", data["Datetime"].iloc[0])
print("To:", data["Datetime"].iloc[-1])

# Fill missing values while retaining timestamps.

data[FEATURES] = data[FEATURES].interpolate(
    method="linear",
    limit_direction="both"
)

data[FEATURES] = data[FEATURES].ffill().bfill()

if data[FEATURES].isna().any().any():
    raise ValueError("Missing values remain after cleaning.")

print("\nMissing values after cleaning:",
      int(data[FEATURES].isna().sum().sum()))

print("Data preprocessing completed!")


# ========================= 5. NORMALIZATION ======================

split_index = int(len(data) * 0.80)

scaler = MinMaxScaler()

# Fit scaler only on training records.

scaler.fit(data[FEATURES].iloc[:split_index])

scaled_data = scaler.transform(data[FEATURES])

print("\nNormalization completed.")


# ========================= 6. SEQUENCE GENERATION ================

X_train, y_train = [], []
X_test, y_test = [], []

# Training sequences

for i in range(LOOKBACK, split_index):

    X_train.append(
        scaled_data[i - LOOKBACK:i]
    )

    y_train.append(
        scaled_data[i, 0]
    )

# Testing sequences

for i in range(split_index, len(data)):

    X_test.append(
        scaled_data[i - LOOKBACK:i]
    )

    y_test.append(
        scaled_data[i, 0]
    )

X_train = np.asarray(X_train, dtype=np.float32)
y_train = np.asarray(y_train, dtype=np.float32)

X_test = np.asarray(X_test, dtype=np.float32)
y_test = np.asarray(y_test, dtype=np.float32)

print("\nTraining Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)


# ========================= 7. BUILD LSTM =========================

model = Sequential([

    Input(shape=(LOOKBACK, len(FEATURES))),

    LSTM(
        32,
        return_sequences=True
    ),

    Dropout(0.2),

    LSTM(16),

    Dropout(0.2),

    Dense(16, activation="relu"),

    Dense(1)

])

model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

print("\nLSTM MODEL SUMMARY")
model.summary()


# ========================= 8. MODEL TRAINING =====================

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

print("\nTraining LSTM model...")

history = model.fit(
    X_train,
    y_train,
    epochs=EPOCHS,
    batch_size=64,
    validation_split=0.1,
    shuffle=False,
    callbacks=[early_stop],
    verbose=1
)

print("\nLSTM Training Completed!")


# ========================= 9. PREDICTION =========================

predicted_scaled = model.predict(
    X_test,
    verbose=0
).flatten()


def inverse_target(values):

    values = np.asarray(values).reshape(-1)

    dummy = np.zeros(
        (len(values), len(FEATURES))
    )

    dummy[:, 0] = values

    return scaler.inverse_transform(dummy)[:, 0]


y_actual = inverse_target(y_test)
y_predicted = inverse_target(predicted_scaled)

print("\nPredictions generated successfully.")


# ========================= 10. MODEL EVALUATION ==================

mae = mean_absolute_error(
    y_actual,
    y_predicted
)

rmse = np.sqrt(
    mean_squared_error(
        y_actual,
        y_predicted
    )
)

print("\n===================================")
print("       WATTWISE AI RESULTS")
print("===================================")

print(f"MAE  : {mae:.5f} kW")
print(f"RMSE : {rmse:.5f} kW")

print("Training records:", len(X_train))
print("Testing records:", len(X_test))

print("===================================")


# ========================= 11. TRAINING GRAPH ====================

plt.figure(figsize=(10, 4))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("LSTM Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig("wattwise_training_loss.png", dpi=150)

plt.show()


# ========================= 12. ACTUAL VS PREDICTED ===============

plt.figure(figsize=(13, 5))

plt.plot(
    y_actual[:500],
    label="Actual Consumption",
    linewidth=2
)

plt.plot(
    y_predicted[:500],
    label="Predicted Consumption",
    linewidth=2
)

plt.title("WattWise AI - Actual vs Predicted Consumption")
plt.xlabel("Test Minute Records")
plt.ylabel("Global Active Power (kW)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig("wattwise_actual_vs_predicted.png", dpi=150)

plt.show()


# ========================= 13. UNSEEN RECORD TEST ================

unseen_sequence = X_test[-1].reshape(
    1, LOOKBACK, len(FEATURES)
)

unseen_scaled = model.predict(
    unseen_sequence,
    verbose=0
).flatten()

unseen_prediction = float(
    inverse_target(unseen_scaled)[0]
)

unseen_actual = float(y_actual[-1])

print("\nUNSEEN TEST RECORD")

print("Predicted:", round(unseen_prediction, 4), "kW")
print("Actual   :", round(unseen_actual, 4), "kW")
print(
    "Absolute Error:",
    round(abs(unseen_prediction - unseen_actual), 4),
    "kW"
)


# ========================= 14. SAVE MODEL ========================

model.save("/content/wattwise_lstm.keras")

import joblib

joblib.dump(
    scaler,
    "/content/wattwise_scaler.pkl"
)

print("\nModel and scaler saved.")


# ========================= 15. DASHBOARD CHARTS ==================

# Dark themed visualization for the website.

def style_chart(ax, fig):

    fig.patch.set_facecolor("#17172D")
    ax.set_facecolor("#20203A")

    ax.tick_params(colors="#D6D0EC")

    ax.xaxis.label.set_color("#C5BEDD")
    ax.yaxis.label.set_color("#C5BEDD")
    ax.title.set_color("#DCCEFF")

    for spine in ax.spines.values():
        spine.set_color("#514A75")

    ax.grid(color="#555071", alpha=0.3)


def comparison_chart():

    fig, ax = plt.subplots(figsize=(10, 4))

    ax.plot(
        y_actual[:500],
        color="#42D7E6",
        linewidth=2,
        label="Actual"
    )

    ax.plot(
        y_predicted[:500],
        color="#B48BFF",
        linewidth=2,
        label="Predicted"
    )

    ax.set_title("Actual vs Predicted Energy Consumption")
    ax.set_xlabel("Minute Records")
    ax.set_ylabel("Active Power (kW)")
    ax.legend(facecolor="#292744", labelcolor="white")

    style_chart(ax, fig)
    fig.tight_layout()

    return fig


def training_chart():

    fig, ax = plt.subplots(figsize=(10, 4))

    ax.plot(
        history.history["loss"],
        color="#42D7E6",
        linewidth=2,
        label="Training"
    )

    ax.plot(
        history.history["val_loss"],
        color="#B48BFF",
        linewidth=2,
        label="Validation"
    )

    ax.set_title("LSTM Training Performance")
    ax.set_xlabel("Epoch")
    ax.set_ylabel("MSE Loss")
    ax.legend(facecolor="#292744", labelcolor="white")

    style_chart(ax, fig)
    fig.tight_layout()

    return fig


def distribution_chart():

    fig, ax = plt.subplots(figsize=(10, 4))

    ax.hist(
        y_actual,
        bins=35,
        color="#42D7E6",
        alpha=0.8
    )

    ax.set_title("Electricity Consumption Distribution")
    ax.set_xlabel("Active Power (kW)")
    ax.set_ylabel("Frequency")

    style_chart(ax, fig)
    fig.tight_layout()

    return fig


# ========================= 16. WEBSITE PREDICTION ================

def website_forecast(file):

    if file is None:
        raise gr.Error("Please upload your electricity dataset.")

    try:

        new_df = pd.read_csv(
            file,
            sep=None,
            engine="python",
            na_values=["?"]
        )

        new_df.columns = new_df.columns.str.strip()

        for col in FEATURES:

            if col not in new_df.columns:
                raise gr.Error(
                    "Required column missing: " + col
                )

            new_df[col] = pd.to_numeric(
                new_df[col],
                errors="coerce"
            )

        if "Date" in new_df.columns and "Time" in new_df.columns:

            new_df["Datetime"] = pd.to_datetime(
                new_df["Date"].astype(str) + " " +
                new_df["Time"].astype(str),
                dayfirst=True,
                errors="coerce"
            )

            new_df = new_df.dropna(
                subset=["Datetime"]
            )

            new_df = new_df.sort_values("Datetime")

        new_df[FEATURES] = (
            new_df[FEATURES]
            .interpolate(
                method="linear",
                limit_direction="both"
            )
            .ffill()
            .bfill()
        )

        if len(new_df) < LOOKBACK:
            raise gr.Error(
                "At least 60 valid records are required."
            )

        if new_df[FEATURES].isna().any().any():
            raise gr.Error("Invalid data detected.")

        latest = new_df[FEATURES].tail(LOOKBACK)

        scaled = scaler.transform(latest)

        sequence = scaled.reshape(
            1,
            LOOKBACK,
            len(FEATURES)
        )

        prediction_scaled = model.predict(
            sequence,
            verbose=0
        ).flatten()

        prediction = float(
            inverse_target(prediction_scaled)[0]
        )

        prediction = max(0, prediction)

        recent_average = float(
            latest[TARGET].mean()
        )

        difference = prediction - recent_average

        trend = (
            "Above recent average"
            if difference > 0
            else "Below recent average"
            if difference < 0
            else "Equal to recent average"
        )

        # Forecast chart

        fig, ax = plt.subplots(figsize=(10, 4))

        ax.plot(
            range(60),
            latest[TARGET].values,
            color="#42D7E6",
            linewidth=2.5,
            label="Historical"
        )

        ax.scatter(
            [60],
            [prediction],
            color="#C19AFF",
            s=180,
            marker="*",
            label="AI Forecast"
        )

        ax.set_title("Next Minute Energy Forecast")
        ax.set_xlabel("Historical Minutes + Forecast")
        ax.set_ylabel("Active Power (kW)")
        ax.legend(facecolor="#292744", labelcolor="white")

        style_chart(ax, fig)
        fig.tight_layout()

        result = f"""
        <div class="forecast-card">
            <div class="small-label">
                AI PREDICTED CONSUMPTION
            </div>

            <div class="big-number">
                {prediction:.4f}
                <span>kW</span>
            </div>

            <div class="forecast-subtitle">
                Next minute electricity consumption
            </div>

            <div class="forecast-divider"></div>

            <div class="data-row">
                <span>Recent 60-minute average</span>
                <b>{recent_average:.4f} kW</b>
            </div>

            <div class="data-row">
                <span>Difference from average</span>
                <b>{difference:+.4f} kW</b>
            </div>

            <div class="data-row">
                <span>Consumption trend</span>
                <b>{trend}</b>
            </div>

            <div class="data-row">
                <span>Records analyzed</span>
                <b>60</b>
            </div>
        </div>
        """

        return result, fig

    except gr.Error:
        raise

    except Exception as error:
        raise gr.Error(str(error))


# ========================= 17. PREMIUM WEBSITE CSS ===============

css = """
@import url('https://fonts.googleapis.com/css2?family=Inter:wght@400;500;600;700;800&display=swap');

.gradio-container {
    background: #101022 !important;
    font-family: 'Inter', sans-serif !important;
    color: #F5F2FF !important;
    max-width: 1550px !important;
    padding: 18px !important;
}

#hero {
    background: linear-gradient(115deg,#282044,#1A3150);
    border: 1px solid #554B78;
    border-radius: 22px;
    padding: 30px;
    margin-bottom: 22px;
    position: relative;
    overflow: hidden;
}

#hero h1 {
    font-size: 43px;
    color: #DCCBFF;
    margin: 12px 0;
    font-weight: 800;
}

#hero p {
    color: #C2BBD9;
}

.hero-tag {
    color: #6EE5EA;
    font-size: 12px;
    letter-spacing: 4px;
    font-weight: 700;
}

.hero-status {
    color: #8CEBD3;
    font-size: 12px;
    margin-top: 20px;
}

#sidebar {
    background: linear-gradient(180deg,#262044,#19182E);
    border: 1px solid #4D456F;
    border-radius: 20px;
    padding: 22px;
    min-height: 680px;
}

#sidebar h2 {
    color: #D5C6FF;
    font-size: 23px;
}

.nav-heading {
    color: #8F89B3;
    font-size: 11px;
    letter-spacing: 2px;
    margin-top: 30px;
}

.nav-item {
    color: #D1CBE6;
    padding: 13px 10px;
    margin-top: 8px;
    border-radius: 10px;
    font-size: 13px;
}

.nav-item.active {
    background: linear-gradient(90deg,#59418F,#294C66);
    color: #FFFFFF;
    border-left: 3px solid #79E2E9;
}

#main-panel {
    background: #17172D;
    border: 1px solid #3D385E;
    border-radius: 20px;
    padding: 22px;
}

.metric-card {
    background: linear-gradient(145deg,#2B2449,#203248);
    border: 1px solid #5A507E;
    border-radius: 17px;
    padding: 22px;
    min-height: 112px;
    transition: 0.3s ease;
}

.metric-card:hover {
    border-color: #8E7AE1;
    transform: translateY(-3px);
}

.metric-label, .small-label {
    color: #AAA4CB;
    font-size: 11px;
    letter-spacing: 1.5px;
}

.metric-value {
    font-size: 28px;
    font-weight: 800;
    color: #E2D4FF;
    margin-top: 12px;
}

.metric-note {
    font-size: 11px;
    color: #75DDE5;
    margin-top: 8px;
}

.section-heading {
    color: #D8C9FF;
    font-size: 23px;
    font-weight: 700;
    margin: 15px 0 8px;
}

.section-description {
    color: #A9A5C4;
    font-size: 13px;
    line-height: 1.8;
}

.forecast-card {
    background: linear-gradient(140deg,#2B214B,#19374A);
    border: 1px solid #6C5A9C;
    border-radius: 18px;
    padding: 25px;
    margin: 10px 0;
}

.big-number {
    color: #8FE9F0;
    font-size: 43px;
    font-weight: 800;
    margin: 15px 0;
}

.big-number span {
    font-size: 19px;
}

.forecast-subtitle {
    color: #C5BFDC;
    font-size: 13px;
}

.forecast-divider {
    border-top: 1px solid #5A527A;
    margin: 20px 0 10px;
}

.data-row {
    display: flex;
    justify-content: space-between;
    gap: 12px;
    padding: 12px 0;
    color: #D2CBE5;
    border-bottom: 1px solid #49415E;
    font-size: 13px;
}

.data-row b {
    color: #8FE9F0;
    text-align: right;
}

button.primary {
    background: linear-gradient(90deg,#865AE5,#2CBBCB) !important;
    border: none !important;
    color: white !important;
    font-weight: 700 !important;
}

button {
    border-radius: 11px !important;
}

.tab-nav button {
    color: #C9C1E5 !important;
}

.tab-nav button.selected {
    color: #8EEBF0 !important;
    border-color: #9676EA !important;
}

label, .prose {
    color: #E0DCF0 !important;
}

footer {
    display: none !important;
}

@media(max-width: 768px) {
    #hero h1 {font-size: 29px;}
    #sidebar {min-height: auto;}
    .metric-value {font-size: 22px;}
}
"""


# ========================= 18. BUILD PREMIUM DASHBOARD ===========

with gr.Blocks(
    theme=gr.themes.Base(
        primary_hue="violet",
        secondary_hue="cyan",
        neutral_hue="slate"
    ),
    css=css,
    title="WattWise AI - Energy Intelligence"
) as demo:

    # HERO HEADER

    gr.HTML("""
    <div id="hero">
        <div class="hero-tag">
            INTELLIGENT ENERGY ANALYTICS PLATFORM
        </div>

        <h1>⚡ WattWise AI</h1>

        <p>
        Predict Tomorrow's Energy. Understand Today's Consumption.
        </p>

        <div class="hero-status">
            ● LSTM MODEL TRAINED &nbsp; | &nbsp;
            DEEP LEARNING POWERED &nbsp; | &nbsp;
            REAL-TIME DEMO
        </div>
    </div>
    """)

    with gr.Row(equal_height=False):

        # SIDEBAR

        with gr.Column(scale=1, min_width=190):

            gr.HTML("""
            <div id="sidebar">

                <h2>⚡ WATTWISE</h2>

                <div class="nav-heading">MAIN MENU</div>

                <div class="nav-item active">
                    ◈ &nbsp; Overview
                </div>

                <div class="nav-item">
                    ◈ &nbsp; Energy Analytics
                </div>

                <div class="nav-item">
                    ◈ &nbsp; Forecast Center
                </div>

                <div class="nav-item">
                    ◈ &nbsp; Model Performance
                </div>

                <div class="nav-item">
                    ◈ &nbsp; About Project
                </div>

                <div class="nav-heading">SYSTEM STATUS</div>

                <div style="
                    background:#203B3A;
                    padding:12px;
                    border-radius:10px;
                    color:#8CEBD3;
                    margin-top:15px;
                    font-size:12px;
                ">
                    ● LSTM MODEL ACTIVE
                </div>

                <div class="nav-heading">MODEL SETTINGS</div>

                <p style="color:#B6B0D1;font-size:12px;
                line-height:2.2;">
                    Historical Window<br>
                    <b style="color:#E2D4FF;">60 Minutes</b>
                    <br>
                    Forecast Interval<br>
                    <b style="color:#8FE9F0;">Next Minute</b>
                </p>

            </div>
            """)

        # MAIN CONTENT

        with gr.Column(scale=4, elem_id="main-panel"):

            gr.HTML("""
            <div class="section-heading">
                Overview Dashboard
            </div>

            <div class="section-description">
                Welcome to WattWise AI. Monitor household
                electricity consumption and deep learning
                forecasting performance.
            </div>
            """)

            # METRIC CARDS

            with gr.Row():

                gr.HTML(f"""
                <div class="metric-card">
                    <div class="metric-label">
                        TEST RECORDS
                    </div>
                    <div class="metric-value">
                        {len(y_actual):,}
                    </div>
                    <div class="metric-note">
                        ✓ Model evaluation
                    </div>
                </div>
                """)

                gr.HTML(f"""
                <div class="metric-card">
                    <div class="metric-label">
                        AVERAGE POWER
                    </div>
                    <div class="metric-value">
                        {np.mean(y_actual):.3f}
                    </div>
                    <div class="metric-note">
                        kW · Test dataset
                    </div>
                </div>
                """)

                gr.HTML(f"""
                <div class="metric-card">
                    <div class="metric-label">
                        MAXIMUM POWER
                    </div>
                    <div class="metric-value">
                        {np.max(y_actual):.3f}
                    </div>
                    <div class="metric-note">
                        kW · Test dataset
                    </div>
                </div>
                """)

            # DASHBOARD NAVIGATION TABS

            with gr.Tabs():

                # OVERVIEW

                with gr.Tab("Overview"):

                    gr.HTML("""
                    <div class="section-heading">
                        Energy Consumption Analytics
                    </div>
                    <div class="section-description">
                        Compare historical actual consumption
                        with LSTM model predictions.
                    </div>
                    """)

                    gr.Plot(
                        value=comparison_chart,
                        label="Actual vs Predicted Consumption"
                    )

                    gr.Plot(
                        value=distribution_chart,
                        label="Consumption Distribution"
                    )

                # ENERGY ANALYTICS

                with gr.Tab("Energy Analytics"):

                    gr.HTML("""
                    <div class="section-heading">
                        Historical Energy Analysis
                    </div>
                    """)

                    gr.Plot(
                        value=distribution_chart,
                        label="Energy Consumption Distribution"
                    )

                    gr.Plot(
                        value=comparison_chart,
                        label="Consumption Comparison"
                    )

                # FORECAST

                with gr.Tab("Forecast Center"):

                    gr.HTML("""
                    <div class="section-heading">
                        ⚡ Intelligent Forecast Center
                    </div>

                    <div class="section-description">
                        Upload household electricity measurements.
                        The LSTM model will analyze the latest
                        60 records and predict the next minute.
                    </div>
                    """)

                    upload_input = gr.File(
                        label="Upload Household Electricity Dataset",
                        file_types=[".txt", ".csv"],
                        type="filepath"
                    )

                    forecast_button = gr.Button(
                        "⚡ Generate Energy Forecast",
                        variant="primary",
                        size="lg"
                    )

                    forecast_result = gr.HTML(
                        label="Prediction Result"
                    )

                    forecast_plot = gr.Plot(
                        label="Forecast Visualization"
                    )

                    forecast_button.click(
                        fn=website_forecast,
                        inputs=upload_input,
                        outputs=[
                            forecast_result,
                            forecast_plot
                        ]
                    )

                # MODEL PERFORMANCE

                with gr.Tab("Model Performance"):

                    gr.HTML("""
                    <div class="section-heading">
                        LSTM Model Performance
                    </div>
                    """)

                    with gr.Row():

                        gr.HTML(f"""
                        <div class="metric-card">
                            <div class="metric-label">
                                MEAN ABSOLUTE ERROR
                            </div>
                            <div class="metric-value">
                                {mae:.5f}
                            </div>
                            <div class="metric-note">
                                kW · Lower is better
                            </div>
                        </div>
                        """)

                        gr.HTML(f"""
                        <div class="metric-card">
                            <div class="metric-label">
                                ROOT MEAN SQUARED ERROR
                            </div>
                            <div class="metric-value">
                                {rmse:.5f}
                            </div>
                            <div class="metric-note">
                                kW · Lower is better
                            </div>
                        </div>
                        """)

                    gr.Plot(
                        value=training_chart,
                        label="Training and Validation Loss"
                    )

                    gr.Plot(
                        value=comparison_chart,
                        label="Actual vs Predicted"
                    )

                # ABOUT

                with gr.Tab("About Project"):

                    gr.HTML("""
                    <div class="section-heading">
                        About WattWise AI
                    </div>

                    <div class="metric-card">

                        <h2 style="color:#D7C7FF;">
                            Smart Energy Intelligence
                        </h2>

                        <p style="
                            color:#D2CDE8;
                            line-height:2;
                        ">
                            WattWise AI is a household electricity
                            consumption forecasting application
                            powered by Long Short-Term Memory.

                            It analyzes historical electricity
                            measurements to predict future
                            consumption and support better
                            energy distribution planning.
                        </p>

                        <hr style="border-color:#51486D;">

                        <h3 style="color:#9AEAF1;">
                            Technology Stack
                        </h3>

                        <p style="color:#D2CDE8;line-height:2;">
                            Python · TensorFlow · Keras · LSTM<br>
                            Pandas · NumPy · Scikit-learn<br>
                            Matplotlib · Gradio
                        </p>

                        <h3 style="color:#9AEAF1;">
                            Model Configuration
                        </h3>

                        <p style="color:#D2CDE8;line-height:2;">
                            Dataset: Household Electric Power Consumption<br>
                            Input Window: 60 Minutes<br>
                            Features: 7 Numerical Measurements<br>
                            Prediction: Next Minute<br>
                            Evaluation: MAE and RMSE
                        </p>

                    </div>
                    """)

    # FOOTER

    gr.HTML("""
    <div style="
        text-align:center;
        padding:30px;
        color:#A6A0C1;
        font-size:12px;
        line-height:2;
    ">
        ⚡ WATTWISE AI
        <br>
        SMART ENERGY INTELLIGENCE
        <br>
        Deep Learning Based Electricity Consumption Forecasting
    </div>
    """)


# ========================= 19. LAUNCH WEBSITE ====================

print("\n" + "=" * 55)
print("WATTWISE AI MODEL TRAINED SUCCESSFULLY")
print("=" * 55)

print("MAE :", round(mae, 5), "kW")
print("RMSE:", round(rmse, 5), "kW")

print("\nLaunching Premium WattWise AI Website...")

demo.launch(
    share=True,
    debug=True,
    show_error=True
)

# ========================= END OF PROJECT =======================
```

```python
%%writefile app.py

import os
import io
import pickle
import numpy as np
import pandas as pd
import gradio as gr
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import tensorflow as tf

MODEL_PATH = "wattwise_lstm.keras"
SCALER_PATH = "wattwise_scaler.pkl"

FEATURES = [
    "Global_active_power",
    "Global_reactive_power",
    "Voltage",
    "Global_intensity",
    "Sub_metering_1",
    "Sub_metering_2",
    "Sub_metering_3"
]

LOOKBACK = 60

model = tf.keras.models.load_model(MODEL_PATH)

with open(SCALER_PATH, "rb") as f:
    scaler = pickle.load(f)


def read_dataset(file_path):
    if not file_path:
        raise gr.Error("Please upload a household electricity TXT or CSV file.")

    df = pd.read_csv(
        file_path,
        sep=";",
        na_values=["?"],
        low_memory=False
    )

    df.columns = df.columns.str.strip()

    missing = [c for c in FEATURES + ["Date", "Time"] if c not in df.columns]
    if missing:
        raise gr.Error("Missing required columns: " + ", ".join(missing))

    df["Datetime"] = pd.to_datetime(
        df["Date"].astype(str) + " " + df["Time"].astype(str),
        dayfirst=True,
        errors="coerce"
    )

    for col in FEATURES:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    df = df.dropna(subset=["Datetime"]).sort_values("Datetime")
    df[FEATURES] = df[FEATURES].interpolate(limit_direction="both")
    df[FEATURES] = df[FEATURES].ffill().bfill()
    df = df.dropna(subset=FEATURES)

    if len(df) < LOOKBACK:
        raise gr.Error("Upload a dataset containing at least 60 valid records.")

    return df


def forecast(file_path):
    try:
        df = read_dataset(file_path)

        latest = df[FEATURES].tail(LOOKBACK).values
        scaled = scaler.transform(latest)
        sequence = scaled.reshape(1, LOOKBACK, len(FEATURES))

        prediction_scaled = model.predict(sequence, verbose=0)[0][0]

        dummy = np.zeros((1, len(FEATURES)))
        dummy[0, 0] = prediction_scaled
        prediction = scaler.inverse_transform(dummy)[0, 0]

        if prediction < 0:
            prediction = 0.0

        last_time = df["Datetime"].iloc[-1]
        next_time = last_time + pd.Timedelta(minutes=1)

        recent = df.tail(60)

        fig, ax = plt.subplots(figsize=(10, 4))
        fig.patch.set_facecolor("#17152b")
        ax.set_facecolor("#211e3b")
        ax.plot(
            recent["Datetime"],
            recent["Global_active_power"],
            color="#43e8f5",
            linewidth=2,
            label="Actual consumption"
        )
        ax.set_title("Recent Electricity Consumption", color="white")
        ax.set_ylabel("Power (kW)", color="white")
        ax.tick_params(colors="#d8d5ef")
        ax.grid(alpha=0.2)
        ax.legend(facecolor="#292544", labelcolor="white")
        fig.autofmt_xdate()
        fig.tight_layout()

        result = f"""
        <div class="result-card">
            <div class="small-label">NEXT MINUTE PREDICTION</div>
            <div class="prediction">{prediction:.4f} <span>kW</span></div>
            <div class="result-note">Forecast time: {next_time}</div>
            <div class="result-note">Based on the latest 60 valid records</div>
        </div>
        """

        return result, fig

    except gr.Error:
        raise
    except Exception as e:
        raise gr.Error(f"Prediction failed: {str(e)}")


css = """
.gradio-container {
    background: #100e20 !important;
    color: #f4f0ff !important;
    max-width: 1200px !important;
}
#hero {
    background: linear-gradient(120deg,#292044,#193451);
    border: 1px solid #514779;
    border-radius: 20px;
    padding: 32px;
    margin-bottom: 22px;
}
#hero h1 { color: #d9c6ff; font-size: 42px; }
#hero p { color: #b8c6f5; }
.panel {
    background: #1b1830;
    border: 1px solid #49416c;
    border-radius: 16px;
    padding: 22px;
}
.result-card {
    background: linear-gradient(120deg,#292044,#17394a);
    border: 1px solid #6554a0;
    border-radius: 16px;
    padding: 25px;
    text-align: center;
}
.small-label {
    color: #55e8ef;
    letter-spacing: 3px;
    font-size: 12px;
}
.prediction {
    color: #e0caff;
    font-size: 42px;
    font-weight: 800;
    margin: 12px;
}
.prediction span { font-size: 18px; color: #55e8ef; }
.result-note { color: #c4c0df; margin-top: 8px; }
button.primary {
    background: linear-gradient(90deg,#8755d9,#267eae) !important;
    border: none !important;
}
"""

with gr.Blocks(
    theme=gr.themes.Base(
        primary_hue="violet",
        secondary_hue="cyan",
        neutral_hue="slate"
    ),
    css=css,
    title="WattWise AI"
) as demo:

    gr.HTML("""
    <div id="hero">
        <p>⚡ INTELLIGENT ENERGY ANALYTICS PLATFORM</p>
        <h1>WattWise AI</h1>
        <p>Understand your electricity consumption.
        Forecast the next minute with deep learning.</p>
        <p>✦ LSTM POWERED &nbsp; | &nbsp;
        ✦ ENERGY ANALYTICS &nbsp; | &nbsp;
        ✦ SMART FORECASTING</p>
    </div>
    """)

    with gr.Tab("Overview"):
        gr.Markdown("## Welcome to WattWise AI")
        gr.Markdown(
            "Upload your household electricity dataset to generate "
            "a next-minute energy consumption prediction."
        )
        gr.Markdown("""
        <div class="panel">
        **MODEL SETTINGS**

        - Model: Long Short-Term Memory (LSTM)
        - Historical window: 60 minutes
        - Forecast interval: Next minute
        - Target: Global active power (kW)
        </div>
        """)

    with gr.Tab("Forecast Center"):
        gr.Markdown("## ⚡ Energy Forecast Center")
        file_input = gr.File(
            label="Upload household_power_consumption.txt or CSV",
            file_types=[".txt", ".csv"],
            type="filepath"
        )
        predict_btn = gr.Button("Generate Energy Forecast", variant="primary")
        output_html = gr.HTML()
        output_plot = gr.Plot(label="Recent Energy Consumption")

        predict_btn.click(
            fn=forecast,
            inputs=file_input,
            outputs=[output_html, output_plot]
        )

    with gr.Tab("About Project"):
        gr.Markdown("""
        ## About WattWise AI

        WattWise AI is a household electricity consumption forecasting
        application built using a Long Short-Term Memory neural network.

        **Technology stack**

        - Python
        - Pandas and NumPy
        - TensorFlow / Keras
        - LSTM Deep Learning
        - Gradio

        **How it works**

        1. Upload household electricity data.
        2. The application reads and cleans the data.
        3. The latest 60 records are passed to the trained LSTM model.
        4. The model predicts the next minute's active power consumption.
        """)

demo.launch(
    server_name="0.0.0.0",
    server_port=int(os.environ.get("PORT", 7860))
)
```
